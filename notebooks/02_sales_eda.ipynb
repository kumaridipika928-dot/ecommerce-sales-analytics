```python
# ============================================
# STEP 15 - RFM ANALYSIS
# ============================================

# Make sure Order Date is a date
cleaned_df["Order Date"] = pd.to_datetime(
    cleaned_df["Order Date"],
    errors="coerce"
)

# Choose a reference date
reference_date = cleaned_df["Order Date"].max() + pd.Timedelta(days=1)

rfm = cleaned_df.groupby("Customer ID").agg({
    "Order Date": lambda x: (reference_date - x.max()).days,
    "Order ID": "nunique",
    "Sales": "sum"
})

# Rename columns
rfm.columns = ["Recency", "Frequency", "Monetary"]

print("RFM Analysis:")
display(rfm.head())
```

```python
# ============================================
# STEP 16 - RFM SCORES
# ============================================

rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    4,
    labels=[4, 3, 2, 1],
    duplicates="drop"
)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    4,
    labels=[1, 2, 3, 4],
    duplicates="drop"
)

# Convert scores to numbers
rfm["R_Score"] = rfm["R_Score"].astype(int)
rfm["F_Score"] = rfm["F_Score"].astype(int)
rfm["M_Score"] = rfm["M_Score"].astype(int)

# Overall RFM score
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

print("RFM scores created!")
display(rfm.head(10))
```

```python


# ============================================
# STEP 17 - CUSTOMER SEGMENTS
# ============================================

def segment_customer(row):

    if row["R_Score"] >= 3 and row["F_Score"] >= 3 and row["M_Score"] >= 3:
        return "VIP Customers"

    elif row["R_Score"] >= 3 and row["F_Score"] >= 2:
        return "Loyal Customers"

    elif row["R_Score"] >= 3:
        return "Recent Customers"

    elif row["R_Score"] <= 2 and row["F_Score"] >= 3:
        return "At Risk Customers"

    else:
        return "Low Engagement"


rfm["Segment"] = rfm.apply(segment_customer, axis=1)

print("Customer Segments:")
display(rfm["Segment"].value_counts())
```

```python
# ============================================
# STEP 18 - CUSTOMER SEGMENT CHART
# ============================================

segment_counts = rfm["Segment"].value_counts()

segment_counts.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Customer Segments")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
from pathlib import Path

# Locate the existing dataset
data_path = Path("../data/raw/sample_-_superstore.xls")

# Load the original data
df = pd.read_excel(data_path)

# Create a separate working copy
cleaned_df = df.copy()

# Standardize column names
cleaned_df.columns = cleaned_df.columns.str.strip()

# Remove duplicate records
cleaned_df = cleaned_df.drop_duplicates().copy()

# Convert date columns into datetime format
for col in ["Order Date", "Ship Date"]:
    if col in cleaned_df.columns:
        cleaned_df[col] = pd.to_datetime(
            cleaned_df[col], errors="coerce"
        )

# Display the result
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", cleaned_df.shape)
print("\nColumns:")
print(cleaned_df.columns.tolist())

cleaned_df.head()
```

```python

# Aggregate sales and profit by product sub-category
subcategory_analysis = (
    cleaned_df.groupby("Sub-Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum")
    )
    .sort_values("Total_Sales", ascending=False)
)

# Calculate profit margin as a percentage of sales
subcategory_analysis["Profit_Margin_%"] = (
    subcategory_analysis["Total_Profit"]
    / subcategory_analysis["Total_Sales"] * 100
).round(2)

display(subcategory_analysis.round(2))
```

```python

import matplotlib.pyplot as plt

subcategory_analysis["Total_Sales"].sort_values().plot(
    kind="barh", figsize=(10, 7)
)

plt.title("Total Sales by Product Sub-Category")
plt.xlabel("Total Sales")
plt.ylabel("Sub-Category")
plt.tight_layout()
plt.show()
```
