```python
# ============================================
# STEP 15 - RFM ANALYSIS
# ============================================

# Make sure Order Date is a date
cleaned_df["Order Date"] = pd.to_datetime(
    cleaned_df["Order Date"],
    errors="coerce"
)

# Choose a reference date
reference_date = cleaned_df["Order Date"].max() + pd.Timedelta(days=1)

rfm = cleaned_df.groupby("Customer ID").agg({
    "Order Date": lambda x: (reference_date - x.max()).days,
    "Order ID": "nunique",
    "Sales": "sum"
})

# Rename columns
rfm.columns = ["Recency", "Frequency", "Monetary"]

print("RFM Analysis:")
display(rfm.head())
```

```python
# ============================================
# STEP 16 - RFM SCORES
# ============================================

rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    4,
    labels=[4, 3, 2, 1],
    duplicates="drop"
)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    4,
    labels=[1, 2, 3, 4],
    duplicates="drop"
)

# Convert scores to numbers
rfm["R_Score"] = rfm["R_Score"].astype(int)
rfm["F_Score"] = rfm["F_Score"].astype(int)
rfm["M_Score"] = rfm["M_Score"].astype(int)

# Overall RFM score
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

print("RFM scores created!")
display(rfm.head(10))
```

```python


# ============================================
# STEP 17 - CUSTOMER SEGMENTS
# ============================================

def segment_customer(row):

    if row["R_Score"] >= 3 and row["F_Score"] >= 3 and row["M_Score"] >= 3:
        return "VIP Customers"

    elif row["R_Score"] >= 3 and row["F_Score"] >= 2:
        return "Loyal Customers"

    elif row["R_Score"] >= 3:
        return "Recent Customers"

    elif row["R_Score"] <= 2 and row["F_Score"] >= 3:
        return "At Risk Customers"

    else:
        return "Low Engagement"


rfm["Segment"] = rfm.apply(segment_customer, axis=1)

print("Customer Segments:")
display(rfm["Segment"].value_counts())
```


### Customer Order Frequency Analysis

Analyzing the number of distinct orders placed by each customer
to understand purchasing frequency and repeat-customer behaviour.


```python

# Count distinct orders placed by each customer
customer_frequency = (
    cleaned_df.groupby(["Customer ID", "Customer Name"])
    .agg(
        Total_Orders=("Order ID", "nunique"),
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .reset_index()
)

print("CUSTOMER ORDER FREQUENCY SUMMARY")
display(
    customer_frequency["Total_Orders"]
    .describe()
    .round(2)
)
```

```python

repeat_customers = (customer_frequency["Total_Orders"] > 1).sum()
one_time_customers = (customer_frequency["Total_Orders"] == 1).sum()
total_customers = len(customer_frequency)

print("CUSTOMER ORDER BEHAVIOUR")
print(f"Total Customers: {total_customers}")
print(f"One-time Customers: {one_time_customers}")
print(f"Repeat Customers: {repeat_customers}")
print(f"Repeat Customer Rate: {repeat_customers / total_customers * 100:.2f}%")
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.hist(
    customer_frequency["Total_Orders"],
    bins=15,
    edgecolor="black"
)

plt.title("Distribution of Orders per Customer")
plt.xlabel("Number of Distinct Orders")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```


### Customer Sales Contribution Analysis

Measuring the contribution of the highest-revenue customers to
overall sales to understand customer concentration and revenue dependence.


```python

# Total sales generated by the top 10 customers
top_10_sales = customer_analysis.nlargest(
    10, "Total_Sales"
)["Total_Sales"].sum()

# Total sales across all transactions
overall_sales = cleaned_df["Sales"].sum()

# Percentage contribution
top_10_contribution = (top_10_sales / overall_sales) * 100

print("CUSTOMER SALES CONTRIBUTION")
print(f"Overall Sales: ${overall_sales:,.2f}")
print(f"Top 10 Customer Sales: ${top_10_sales:,.2f}")
print(f"Top 10 Contribution: {top_10_contribution:.2f}%")
```

```python

import matplotlib.pyplot as plt

other_customer_sales = overall_sales - top_10_sales

plt.figure(figsize=(7, 5))
plt.bar(
    ["Top 10 Customers", "All Other Customers"],
    [top_10_sales, other_customer_sales]
)

plt.title("Sales Contribution: Top 10 vs Other Customers")
plt.ylabel("Total Sales ($)")
plt.tight_layout()
plt.show()
```

```python

top_10_pct = top_10_sales / overall_sales * 100
other_pct = 100 - top_10_pct

plt.figure(figsize=(8, 5))

bars = plt.bar(
    ["Top 10 Customers", "All Other Customers"],
    [top_10_pct, other_pct]
)

plt.bar_label(bars, fmt="%.2f%%", padding=4)
plt.title("Share of Total Sales by Customer Group")
plt.ylabel("Share of Total Sales (%)")
plt.ylim(0, 110)
plt.tight_layout()
plt.show()
```


## 5. Sales Trends Analysis

Analyzing monthly and yearly sales performance to identify
growth patterns, fluctuations, and potential seasonality.


```python

# Aggregate sales by month
monthly_sales = (
    cleaned_df.set_index("Order Date")
    .resample("ME")["Sales"]
    .sum()
)

plt.figure(figsize=(12, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

monthly_profit = (
    cleaned_df.set_index("Order Date")
    .resample("ME")["Profit"]
    .sum()
)

plt.figure(figsize=(12, 5))
plt.plot(monthly_profit.index, monthly_profit.values, marker="o")

plt.axhline(0, linewidth=1)
plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Total Profit ($)")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

yearly_performance = (
    cleaned_df.groupby(cleaned_df["Order Date"].dt.year)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Orders=("Order ID", "nunique")
    )
)

yearly_performance["Profit_Margin_%"] = (
    yearly_performance["Total_Profit"]
    / yearly_performance["Total_Sales"] * 100
)

print("YEARLY PERFORMANCE SUMMARY")
display(yearly_performance.round(2))
```
